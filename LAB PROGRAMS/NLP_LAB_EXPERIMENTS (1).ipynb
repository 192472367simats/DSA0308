```python
def finite_automaton(string):
    state = 0

    for char in string:
        if state == 0:
            if char == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if char == 'b':
                state = 2
            elif char == 'a':
                state = 1
            else:
                state = 0

        elif state == 2:
            if char == 'a':
                state = 1
            else:
                state = 0

    return state == 2

strings = ["ab", "aab", "abab", "abc", "baaab"]

for s in strings:
    if finite_automaton(s):
        print(s, "Accepted")
    else:
        print(s, "Rejected")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet

# Download required resources (Run only once)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()

# Input sentence
text = "The boys are running, the girls are writing letters, and the children played games."

# Tokenize the sentence
words = word_tokenize(text)

# Perform POS tagging
pos_tags = nltk.pos_tag(words)

# Function to convert NLTK POS tags to WordNet POS
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

print("Word\t\tPOS\tLemma")
print("-" * 35)

# Perform Morphological Analysis
for word, tag in pos_tags:
    lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    print(f"{word:12}{tag:8}{lemma}")
```

```python
def generate_plural(noun):

    if noun.endswith(("s", "x", "z", "ch", "sh")):
        plural = noun + "es"

    elif noun.endswith("y") and len(noun) > 1 and noun[-2].lower() not in "aeiou":
        plural = noun[:-1] + "ies"

    else:
        plural = noun + "s"

    return plural

words = [
    "cat", "dog", "bus", "box", "church",
    "dish", "baby", "city", "toy", "book"
]

print("Singular\tPlural")
print("------------------------")

for word in words:
    print(f"{word:10}\t{generate_plural(word)}")
```

```python
import nltk
from nltk.util import bigrams
from collections import defaultdict
import random

nltk.download('punkt')
nltk.download('punkt_tab')

text = "Natural language processing is an exciting field of artificial intelligence."

tokens = nltk.word_tokenize(text)

bigram_pairs = list(bigrams(tokens))

model = defaultdict(list)

for w1, w2 in bigram_pairs:
    model[w1].append(w2)

word = tokens[0]
generated = [word]

for i in range(10):
    if word in model:
        word = random.choice(model[word])
        generated.append(word)
    else:
        break

print("Generated Text:")
print(" ".join(generated))
```

```python
class State:
    def __init__(self, lhs, rhs, dot, start):
        self.lhs = lhs
        self.rhs = rhs
        self.dot = dot
        self.start = start

    def is_complete(self):
        return self.dot == len(self.rhs)

    def next_symbol(self):
        if self.dot < len(self.rhs):
            return self.rhs[self.dot]
        return None

    def advance(self):
        return State(self.lhs, self.rhs, self.dot + 1, self.start)

    def __eq__(self, other):
        return (self.lhs, self.rhs, self.dot, self.start) == \
               (other.lhs, other.rhs, other.dot, other.start)

    def __hash__(self):
        return hash((self.lhs, tuple(self.rhs), self.dot, self.start))

    def __repr__(self):
        rhs = self.rhs[:]
        rhs.insert(self.dot, "•")
        return f"{self.lhs} -> {' '.join(rhs)}, ({self.start})"


def earley_parse(grammar, sentence, start_symbol):
    words = sentence.split()
    n = len(words)

    chart = [set() for _ in range(n + 1)]
    chart[0].add(State("γ", [start_symbol], 0, 0))

    for i in range(n + 1):
        changed = True

        while changed:
            changed = False
            states = list(chart[i])

            for state in states:

                next_sym = state.next_symbol()

                # Predictor
                if next_sym in grammar:
                    for production in grammar[next_sym]:
                        new_state = State(next_sym, production, 0, i)

                        if new_state not in chart[i]:
                            chart[i].add(new_state)
                            changed = True

                # Scanner
                elif next_sym is not None and i < n:
                    if next_sym == words[i]:
                        new_state = state.advance()

                        if new_state not in chart[i + 1]:
                            chart[i + 1].add(new_state)

                # Completer
                elif state.is_complete():
                    for old_state in list(chart[state.start]):
                        if old_state.next_symbol() == state.lhs:
                            new_state = old_state.advance()

                            if new_state not in chart[i]:
                                chart[i].add(new_state)
                                changed = True

    final_state = State("γ", [start_symbol], 1, 0)

    if final_state in chart[n]:
        print("Sentence Accepted")
    else:
        print("Sentence Rejected")


grammar = {
    "S": [["NP", "VP"]],
    "NP": [["John"], ["Mary"]],
    "VP": [["V", "NP"]],
    "V": [["likes"], ["sees"]]
}

sentence = "John likes Mary"

earley_parse(grammar, sentence, "S")
```

```python
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> DET N
VP -> V NP
DET -> 'THE'
N -> 'CAT' | 'DOG'
V -> 'CHASED'
""")

parser = EarleyChartParser(grammar)

sentence = "THE CAT CHASED THE DOG".split()

trees = list(parser.parse(sentence))

if trees:
    print("Sentence is acceptable")

    for tree in trees:
        tree.pretty_print()
else:
    print("Sentence is not acceptable")
```

```python
grammar = {

    "S": ["NP VP"],

    "NP": ["DET N"],

    "VP": ["V"],

    "DET": ["the"],

    "N singular": ["boy", "girl", "student"],

    "N plural": ["boys", "girls", "students"],

    "V singular": ["runs", "plays", "eats"],

    "V plural": ["run", "play", "eat"]

}


def check(sentence):

    words = sentence.lower().split()

    if len(words) != 3:
        return "Sentence must contain only 3 words"

    determiner, noun, verb = words

    if noun in grammar["N singular"] and verb in grammar["V singular"]:
        return "Sentence is accepted"

    if noun in grammar["N plural"] and verb in grammar["V plural"]:
        return "Sentence is accepted"

    return "Sentence is not accepted"


sentences = [
    "the girl runs",
    "the girls run",
    "the girls runs",
    "the students eat"
]

for sentence in sentences:
    print(sentence, "->", check(sentence))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "Natural language processing is interesting",
    "Machine learning is useful",
    "Natural language processing uses machine learning"
]

query = "natural language processing"

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(documents)

query_vector = vectorizer.transform([query])

scores = (tfidf @ query_vector.T).toarray().flatten()

print("Document Ranking:")

for i in scores.argsort()[::-1]:
    print("Document", i + 1, "Score:", round(scores[i], 3))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = """
Natural language processing deals with human language.
NLP uses machine learning techniques.
Machine learning helps computers understand text.
"""

sentences = [s.strip() for s in text.strip().split(".") if s.strip()]

vectorizer = TfidfVectorizer()

vectors = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(vectors)

score = 0

for i in range(len(sentences) - 1):
    score += similarity[i][i + 1]

score = score / (len(sentences) - 1)

print("Coherence Score:", round(score, 3))

if score > 0.1:
    print("Text is coherent")
else:
    print("Text is less coherent")
```

```python
def dialogue_act(sentence):

    s = sentence.lower()

    if "?" in s:
        return "QUESTION"

    elif s.startswith(("please", "can you", "could you")):
        return "REQUEST"

    elif s.startswith(("hello", "hi", "good morning")):
        return "GREETING"

    elif s.startswith(("thank", "thanks")):
        return "THANKING"

    elif s.startswith(("yes", "okay", "ok")):
        return "ACCEPTANCE"

    else:
        return "STATEMENT"


dialogues = [
    "Hello!",
    "Can you help me?",
    "Please open the door.",
    "Thank you.",
    "The class starts at 9 AM."
]

for sentence in dialogues:
    print(sentence, "->", dialogue_act(sentence))
```
